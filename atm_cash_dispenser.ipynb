## atm cash dispenses

```python
def atm_cash_dispenser(amount):
    denominations = [2000, 500,200, 100, 50, 20, 10]
    for note in denominations:
        count = amount // note
        if count:
            yield f"{count} x ₹{note}"
        amount %= note
    if amount > 0:
        yield "Remaining amount cannot be dispensed."

# Example Usage
withdrawal_amount = int(input("Enter amount to withdraw: "))  # User input
print("Cash Dispensal:")
for cash in atm_cash_dispenser(withdrawal_amount):
    print(cash)
```
