## movie ticket booking

```python
def seat_number_generator(row='A', seats=10):
    for i in range(1, seats + 1):
        yield f"Seat {row}{i}"

# Example Usage
row_letter = input("Enter row letter: ")  # User input
num_seats = int(input("Enter number of seats: "))  # User input

seats = seat_number_generator(row_letter, num_seats)
for _ in range(num_seats):
    print(next(seats))
```
