## wakeup_remainder -->wakeup alert(morning routine notification)

```python
import time

def wakeup_reminder(func):
    def wrapper(*args, **kwargs):
        print("Reminder: It's time to wake up! ")
        return func(*args, **kwargs)
    return wrapper


@wakeup_reminder
def morning_routine(name):
    print(f"Good morning, {name}! Time to start your day.")

name = input("Enter your name: ")

morning_routine(name)
```
