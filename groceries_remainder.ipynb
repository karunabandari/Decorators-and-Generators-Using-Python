## groceries_remainder--> prevents forgetting items(groceries checklist)

```python
def grocery_reminder(func):
    def wrapper(*args, **kwargs):
        print("Reminder: Check your grocries list before leaving!")
        return func(*args, **kwargs)
    return wrapper

@grocery_reminder
def grocery_list(items):
    print("Groceries list:", ", ".join(items))
    print("Thank you visit again....")

# Input
items = input("Enter grocery items (comma-separated): ").split(",")

grocery_list(items)
```
